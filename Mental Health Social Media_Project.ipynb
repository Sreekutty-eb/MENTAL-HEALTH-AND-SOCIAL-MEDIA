```python
from sklearn.preprocessing import LabelEncoder

obj1=LabelEncoder()
obj2=LabelEncoder()
obj3=LabelEncoder()
obj4=LabelEncoder()
obj5=LabelEncoder()

data['person_name']=obj1.fit_transform(data['person_name'])
data['date']=obj2.fit_transform(data['date'])
data['gender']=obj3.fit_transform(data['gender'])
data['platform']=obj4.fit_transform(data['platform'])
data['mental_state']=obj5.fit_transform(data['mental_state'])
```
